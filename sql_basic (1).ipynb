#1 Create a new database and table for employees


```python
import sqlite3

# Connect to the SQLite database (creates the database file if it doesn't exist)
conn = sqlite3.connect('company.db')
cursor = conn.cursor()

# Create the employees table
cursor.execute('''
CREATE TABLE IF NOT EXISTS employees(
    employee_id INTEGER PRIMARY KEY,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    department VARCHAR(50),
    salary INT,
    hire_date DATE
);
''')

# Commit the changes and close the connection
conn.commit()
conn.close()

print("Database 'company.db' and table 'employees' created successfully.")
```

#3 Display all employees records sorted by salary (lowest to highest)


```python
import sqlite3

# Connect to the SQLite database
conn = sqlite3.connect('company.db')
cursor = conn.cursor()

# Select all employees records sorted by salary (lowest to highest)
cursor.execute('''
SELECT *
FROM employees
ORDER BY salary ASC;
''')

# Fetch all the results
employees = cursor.fetchall()

# Print the results
for employee in employees:
    print(employee)

# Close the connection
conn.close()
```

#4 Show employees Sorted by Department (A-Z and salary(High to low)


```python
import sqlite3

# Connect to the SQLite database
conn = sqlite3.connect('company.db')
c = conn.cursor()

# Select employees sorted by department (A-Z) and then by salary (High to low)
c.execute('''
SELECT *
FROM employees
ORDER BY department ASC, salary DESC;
''')

# Fetch all the results
employees_sorted = c.fetchall()

# Print the results
for employee in employees_sorted:
    print(employee)

# Close the connection
conn.close()
```

#5 List all employees in the IT department,ordered by hire date (Newest First)


```python
import sqlite3

# Connect to the SQLite database
conn = sqlite3.connect('company.db')
cursor = conn.cursor()

# List all employees in the IT department, ordered by hire date (Newest First)
cursor.execute('''
SELECT *
FROM employees
WHERE department='IT'
ORDER BY hire_date DESC;
''')

# Fetch all the results
it_employees = cursor.fetchall()

# Print the results
for employee in it_employees:
    print(employee)

# Close the connection
conn.close()
```

#6 Create and populates a sales tables


```python
import sqlite3

# Connect to the SQLite database
conn = sqlite3.connect('company.db')
cursor = conn.cursor()

# Create the sales table
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales(
    sale_id INTEGER PRIMARY KEY,
    customer_name VARCHAR(50),
    amount INT,
    sale_date DATE
);
''')

# Data to be inserted into the sales table
sales_data = [
    (1,'Aditi',1500,'2024-08-01'),
    (2,'Rohan',2200,'2024-08-03'),
    (3,'Aditi',3500,'2024-09-05'),
    (4,'Meena',2700,'2024-09-15'),
    (5,'Rohan',4500,'2024-09-25')
]

# Insert multiple data rows into the sales table
cursor.executemany('''
INSERT INTO sales (sale_id, customer_name, amount, sale_date)
VALUES (?, ?, ?, ?)
''', sales_data)

# Commit the changes and close the connection
conn.commit()
conn.close()

print("Sales table created and data inserted successfully.")
```

#7 Display all sales records sorted by amount (highest-lowest)


```python
import sqlite3

# Connect to the SQLite database
conn = sqlite3.connect('company.db')
cursor = conn.cursor()

# Select all sales records sorted by amount (highest-lowest)
cursor.execute('''
SELECT *
FROM sales
ORDER BY amount DESC;
''')

# Fetch all the results
sales_records = cursor.fetchall()

# Print the results
for sale in sales_records:
    print(sale)

# Close the connection
conn.close()
```

#8 Show all sales made by customer ''Aditi''


```python
import sqlite3

# Connect to the SQLite database
conn = sqlite3.connect('company.db')
cursor = conn.cursor()

# Show all sales made by customer 'Aditi'
cursor.execute('''
SELECT *
FROM sales
WHERE customer_name='Aditi'
''')

# Fetch all the results
aditi_sales = cursor.fetchall()

# Print the results
for sale in aditi_sales:
    print(sale)

# Close the connection
conn.close()
```
